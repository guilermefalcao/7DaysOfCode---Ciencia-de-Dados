# Dia 1 - Exploração CEAPS (CORRIGIDO)

Vamos explorar os dados de despesas dos senadores!

## 1. Importar Bibliotecas

```python
import pandas as pd
import numpy as np

print("✅ Bibliotecas importadas!")
```

## 2. Carregar Dados

```python
# Carregar pulando a primeira linha (cabeçalho com data)
df = pd.read_csv('../Dados Abertos - CEAPS - csv/despesa_ceaps_2022.csv', 
                 encoding='latin-1', 
                 sep=';',
                 skiprows=1)  # Pula a linha com 'ULTIMA ATUALIZACAO'

print(f"📊 Dados carregados: {df.shape[0]} linhas e {df.shape[1]} colunas")
```

## 3. VER NOMES DAS COLUNAS (IMPORTANTE!)

```python
print("📋 NOMES EXATOS DAS COLUNAS:")
print("="*50)
for i, col in enumerate(df.columns, 1):
    print(f"{i}. '{col}'")
print("="*50)
```

## 4. Primeiras Linhas

```python
print("🔍 Primeiras 3 linhas:")
df.head(3)
```

## 5. Limpar Nomes das Colunas

```python
# Remover espaços e aspas dos nomes das colunas
df.columns = df.columns.str.strip().str.replace('"', '')

print("✅ Colunas limpas!")
print("\nNovos nomes:")
print(list(df.columns))
```

## 6. Estatísticas dos Senadores

```python
print("👥 Quantos senadores diferentes?")
print(f"Total: {df['SENADOR'].nunique()} senadores")

print("\n🔝 Top 10 senadores com mais despesas:")
top_senadores = df['SENADOR'].value_counts().head(10)
print(top_senadores)
```

## 7. Tipos de Despesas

```python
print("💰 Tipos de despesas:")
tipos = df['TIPO_DESPESA'].value_counts()
print(tipos)
```

## 8. Análise de Valores

```python
# Converter valores para numérico
df['VALOR_NUM'] = df['VALOR_REEMBOLSADO'].str.replace(',', '.').astype(float)

print("💵 Estatísticas dos valores:")
print(f"Valor mínimo: R$ {df['VALOR_NUM'].min():.2f}")
print(f"Valor máximo: R$ {df['VALOR_NUM'].max():.2f}")
print(f"Valor médio: R$ {df['VALOR_NUM'].mean():.2f}")
print(f"Valor total: R$ {df['VALOR_NUM'].sum():,.2f}")
```

## 9. Top 10 Senadores por Valor Gasto

```python
print("💰 Top 10 senadores que mais gastaram:")
top_gastos = df.groupby('SENADOR')['VALOR_NUM'].sum().sort_values(ascending=False).head(10)
print(top_gastos)
```

## 10. Despesas por Mês

```python
print("📅 Total gasto por mês em 2022:")
despesas_mes = df.groupby('MES')['VALOR_NUM'].sum().sort_index()
for mes, valor in despesas_mes.items():
    print(f"Mês {mes}: R$ {valor:,.2f}")
```

## 11. Resumo Final

```python
print("="*60)
print("📊 RESUMO COMPLETO - CEAPS 2022")
print("="*60)
print(f"Total de registros: {len(df):,}")
print(f"Total de senadores: {df['SENADOR'].nunique()}")
print(f"Total de fornecedores: {df['FORNECEDOR'].nunique()}")
print(f"Total de tipos de despesa: {df['TIPO_DESPESA'].nunique()}")
print(f"\nValor total gasto: R$ {df['VALOR_NUM'].sum():,.2f}")
print(f"Média por despesa: R$ {df['VALOR_NUM'].mean():.2f}")
print(f"Maior despesa: R$ {df['VALOR_NUM'].max():,.2f}")
print(f"Menor despesa: R$ {df['VALOR_NUM'].min():.2f}")
print("="*60)
```
