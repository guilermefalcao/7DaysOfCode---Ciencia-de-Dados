## 5. Consolidar Dados de Múltiplos Anos

```python
# Carregar e consolidar dados de 2017 a 2022
anos = range(2017, 2023)
dfs = []

for ano in anos:
    df = pd.read_csv(f'../Dados Abertos - CEAPS - csv/despesa_ceaps_{ano}.csv', encoding='latin-1', sep=';')
    df['ANO'] = ano
    dfs.append(df)
    print(f"Ano {ano}: {df.shape[0]} registros")

# Concatenar todos os dataframes
df_completo = pd.concat(dfs, ignore_index=True)
print(f"\nDataset completo: {df_completo.shape}")
```
